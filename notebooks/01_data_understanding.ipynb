```python
from pathlib import Path
import pandas as pd
import numpy as np

# Project root
ROOT_DIR = Path.cwd().parent

# Data directory
DATA_DIR = ROOT_DIR / "data" / "raw"

print("Project root :", ROOT_DIR)
print("Data folder  :", DATA_DIR)
```

```python
files = {
    "customers": "olist_customers_dataset.csv",
    "geolocation": "olist_geolocation_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "order_payments": "olist_order_payments_dataset.csv",
    "order_reviews": "olist_order_reviews_dataset.csv",
    "orders": "olist_orders_dataset.csv",
    "products": "olist_products_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "category_translation": "product_category_name_translation.csv"
}

dfs = {}

for name, filename in files.items():
    path = DATA_DIR / filename
    
    dfs[name] = pd.read_csv(path)
    
    print(f"{name:20} {dfs[name].shape}")
```

```python
def profile_dataframe(df):
    profile = pd.DataFrame({
        "column": df.columns,
        "dtype": df.dtypes.astype(str),
        "missing_count": df.isna().sum().values,
        "missing_pct": (df.isna().mean() * 100).round(2).values,
        "unique_count": df.nunique().values
    })
    
    return profile
```

```python
def check_key(df, column):
    return {
        "column": column,
        "null_count": df[column].isna().sum(),
        "unique_count": df[column].nunique(),
        "row_count": len(df),
        "duplicate_count": df[column].duplicated().sum()
    }
```

```python
key_checks = pd.DataFrame([
    check_key(dfs["customers"], "customer_id"),
    check_key(dfs["orders"], "order_id"),
    check_key(dfs["products"], "product_id"),
    check_key(dfs["sellers"], "seller_id")
])

key_checks
```

```python
orders = dfs["orders"].copy()

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(
        orders[col],
        errors="coerce"
    )
```

```python
order_items = dfs["order_items"].copy()

sales = orders.merge(
    order_items,
    on="order_id",
    how="left"
)
```

```python
sales.shape
```

```python
total_revenue = sales["price"].sum()

print(f"Total Revenue: R$ {total_revenue:,.2f}")
```

```python
total_freight = sales["freight_value"].sum()

print(f"Total Freight: R$ {total_freight:,.2f}")
```

```python
gmv = sales["price"].sum() + sales["freight_value"].sum()

print(f"GMV: R$ {gmv:,.2f}")
```

```python
sales["purchase_month"] = (
    sales["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
monthly_sales = (
    sales.groupby("purchase_month")
    .agg(
        revenue=("price", "sum"),
        orders=("order_id", "nunique"),
        items=("order_item_id", "count")
    )
    .reset_index()
)

monthly_sales.head()
```

```python
monthly_sales["aov"] = (
    monthly_sales["revenue"]
    / monthly_sales["orders"]
)

monthly_sales.head()
```

```python
delivered_orders = orders[
    orders["order_status"] == "delivered"
].copy()
```

```python
delivered_orders["delivery_days"] = (
    delivered_orders["order_delivered_customer_date"]
    - delivered_orders["order_purchase_timestamp"]
).dt.total_seconds() / (60 * 60 * 24)
```

```python
delivered_orders["delivery_delay_days"] = (
    delivered_orders["order_delivered_customer_date"]
    - delivered_orders["order_estimated_delivery_date"]
).dt.total_seconds() / (60 * 60 * 24)
```

```python
delivered_orders["is_late"] = (
    delivered_orders["delivery_delay_days"] > 0
)
```

```python
avg_delivery_days = delivered_orders["delivery_days"].mean()

late_delivery_rate = delivered_orders["is_late"].mean() * 100

print(f"Average Delivery Days : {avg_delivery_days:.2f}")
print(f"Late Delivery Rate    : {late_delivery_rate:.2f}%")
```

```python
PROCESSED_DIR = ROOT_DIR / "data" / "processed"

PROCESSED_DIR.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
sales.to_csv(
    PROCESSED_DIR / "fact_sales.csv",
    index=False
)

delivered_orders.to_csv(
    PROCESSED_DIR / "fact_delivery.csv",
    index=False
)

monthly_sales.to_csv(
    PROCESSED_DIR / "monthly_sales.csv",
    index=False
)
```
